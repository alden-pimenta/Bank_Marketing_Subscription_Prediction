```python
df = pd.read_csv("bank-additional-full.csv",sep=";")
df.head(3)
```

```python
import numpy as np

df["not_previously_contacted"] = (df["pdays"] == 999).astype(int)
df["pdays"] = df["pdays"].replace(999, np.nan)
```

```python
# Check the values
print(df["pdays"].value_counts(dropna=False).head())
```

```python
df["pdays"] = df["pdays"].fillna(df["pdays"].median())
```

```python
df = pd.get_dummies(df, columns=["marital" ,"default", "housing","loan","contact","poutcome"],drop_first=True, dtype=int)
```

```python
df.sample(5)
```

```python
df["job"] = df['job'].map({
    'housemaid': 0,
    'services': 1,
    'admin.': 2,
    'blue-collar': 3,
    'technician': 4,
    'retired': 5,
    'management': 6,
    'unemployed': 7,
    'self-employed': 8,
    'unknown': 9,
    'entrepreneur': 10,
    'student': 11
})
```

```python
df["education"].unique()
```

```python
df["education"] = df["education"].map({
    "unknown": 0,
    "illiterate": 1,
    "basic.4y": 2,
    "basic.6y": 3,
    "basic.9y": 4,
    "high.school": 5,
    "professional.course": 6,
    "university.degree": 7
})
```

```python
df.head(3)
```

```python
df.drop(["month","day_of_week","duration"],axis=1, inplace=True)
```

```python
df.head()
```

```python
df["y"]= df["y"].map({"yes":1, "no":0})
```

```python
X = df.drop("y",axis=1)
y = df["y"]

y.sample(5)
X.sample(5)
```

<h3 style="color:#7C0E44; border: 1px solid #7C0E44; padding:5px; text-align:center">Train Test Split </h3>


```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.2, random_state=40)
```

```python
print(X_train.shape)
print(X_test.shape)
```

```python
X_train.head(3)
```

```python
y_train.head(3)
```

<h3 style="color:#7C0E44; border: 1px solid #7C0E44; padding:4px; text-align:center">Feature Sandardization</h3>

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

X_train = pd.DataFrame(  X_train_scaled, columns=X_train.columns)
X_test = pd.DataFrame(  X_test_scaled, columns=X_test.columns)

X_train.head(3)
```

<h3 style="color:#7C0E44; border: 1px solid #7C0E44; padding:4px; text-align:center">Applying SMOTE</h3>

```python
from imblearn.over_sampling import SMOTE

smote = SMOTE()
X_train_resampled, y_train_resampled = smote.fit_resample(
    X_train, y_train
)
```

```python
X_train_resampled.sample(5)
```

<h3 style="color:#7C0E44; border: 2px solid #7C0E44; padding:5px; text-align:center">Model Building </h3>


<h3 style="color:#7C0E44; border: 1px solid #7C0E44; padding:4px; text-align:center">Logistic Regression</h3>

```python
from sklearn.linear_model import LogisticRegression

lr_model = LogisticRegression()
lr_model.fit(X_train_resampled, y_train_resampled)
```

```python
print(lr_model.score(X_test, y_test))
```

```python
from sklearn.metrics import classification_report , confusion_matrix

y_pred_lr = lr_model.predict(X_test)
print(classification_report(y_test, y_pred_lr))
```

```python
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred_lr))
```

<h3 style="color:#7C0E44; border: 1px solid #7C0E44; padding:5px; text-align:center">Naive Bayes </h3>

```python
from sklearn.naive_bayes import BernoulliNB

nb_model = BernoulliNB()
nb_model.fit(X_train_resampled, y_train_resampled)
```

```python
nb_model.score(X_test, y_test)
```

```python
y_pred_nb = nb_model.predict(X_test)
print(classification_report(y_test, y_pred_nb))
```

```python
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred_nb))
```

<h3 style="color:#7C0E44; border: 1px solid #7C0E44; padding:5px; text-align:center">SVM</h3>

```python
from sklearn.svm import LinearSVC

svm_model = LinearSVC(class_weight="balanced", random_state=40, max_iter=5000)
svm_model.fit(X_train_resampled, y_train_resampled)
```

```python
y_pred_svm = svm_model.predict(X_test)
print(classification_report(y_test, y_pred_svm, zero_division=0))
```

```python
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred_svm))
```

<h3 style="color:#7C0E44; border: 1px solid #7C0E44; padding:5px; text-align:center">Cross Validation for KNN </h3>

```python
# Cross Validation
from sklearn.neighbors import KNeighborsClassifier
from imblearn.pipeline import Pipeline
from sklearn.model_selection import GridSearchCV

pipe = Pipeline([
    ("scale", StandardScaler()),
    ("smote", SMOTE(random_state=40)),
    ("knn", KNeighborsClassifier())
])
param_grid = {"knn__n_neighbors":[ i for i in range(1,30,2)]}

classifier_cv = GridSearchCV(
    pipe,
    param_grid,
    cv = 5,
    scoring="f1"
)

classifier_cv.fit(X_train, y_train)
```

```python
from sklearn.metrics import classification_report

y_pred_cv = classifier_cv.predict(X_test)
print(classification_report(y_test, y_pred_cv))

print("\n")

```

```python
classifier_cv.best_params_
```

```python
knn_model = KNeighborsClassifier(
    n_neighbors=classifier_cv.best_params_['knn__n_neighbors']
)

knn_model.fit(X_train_resampled, y_train_resampled)
```

```python
from sklearn.metrics import classification_report

y_pred_knn = knn_model.predict(X_test)
print(classification_report(y_test,y_pred_knn))
```

```python

print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred_knn))
```

<h3 style="color:#7C0E44; border: 1px solid #7C0E44; padding:5px; text-align:center">Random Forest</h3>

<h3 style="color:#7C0E44; border: 1px solid #7C0E44; padding:5px; text-align:center">Gradient Boosting Algorithm </h3>

```python
from sklearn.ensemble import GradientBoostingClassifier

gb_model = GradientBoostingClassifier(
    n_estimators=25, learning_rate=0.01, max_depth=10, random_state=40,min_samples_split=5 ,min_samples_leaf=20
)
gb_model.fit(X_train_resampled, y_train_resampled)
```

```python
y_pred_gb = gb_model.predict(X_test)
print(classification_report(y_test, y_pred_gb))
```

<h3 style="color:#7C0E44; border: 1px solid #7C0E44; padding:5px; text-align:center">Evaluation</h3>

```python
from sklearn.metrics import accuracy_score, precision_score, f1_score, recall_score

model_predictions = {
    "Logistic Regression": y_pred_lr,
    "Naive Bayes": y_pred_nb,
    "Linear SVM": y_pred_svm,
    "KNN": y_pred_knn,
    "Decision Tree": y_pred_DT,
    "Random Forest": y_pred_RF,
    "Gradient Boosting": y_pred_gb
}

results = []

for name, pred in model_predictions.items():
    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, pred),
        "Precision": precision_score(y_test, pred, zero_division=0),
        "Recall": recall_score(y_test, pred, zero_division=0),
        "F1 Score": f1_score(y_test, pred, zero_division=0)
    })

model_results = pd.DataFrame(results).sort_values(
    "F1 Score", ascending=False
).reset_index(drop=True)

model_results.round(3)
```

```python
# Compare model scores visually
model_results.set_index("Model")[["Accuracy", "Precision", "Recall", "F1 Score"]].plot(
    kind="bar", figsize=(12,5)
)
plt.title("Model Performance Comparison")
plt.ylabel("Score")
plt.ylim(0, 1)
plt.style.use("classic")
plt.xticks(rotation=35, ha="right")
plt.tight_layout()
plt.show()
```

### 1. Objective

The objective of this project is to predict whether a customer will subscribe to a term deposit based on their personal details, previous interactions and marketing campaign information. This can help the bank identify potential customers and improve its marketing strategy.

---

### 2. Data Analysis

The dataset contains 41,188 records and 21 columns, including customer information, campaign details, economic indicators and the target variable y. Exploratory Data Analysis (EDA) was performed to understand customer behaviour, feature distributions, unknown values and class imbalance.

---

### 3. Data Preprocessing

Unnecessary columns were removed, and missing values and duplicate records were checked. Categorical features were converted into numerical values, and feature scaling was performed using `StandardScaler`. SMOTE was applied to the training data to address class imbalance and help the models identify potential subscribers.

---

### 4. Model Building and Evaluation

Seven classification algorithms were evaluated: Logistic Regression, Gaussian Naive Bayes, Linear SVM, KNN, Decision Tree, Random Forest, and Gradient Boosting.

The models were compared using accuracy, precision, recall, and F1-score. Since most customers did not subscribe, accuracy alone was not considered sufficient for selecting the best model.

---

### 5. Model Comparison Report

Seven machine learning classification models were evaluated to predict whether a bank customer would subscribe to a term deposit.

#### Performance Comparison

| Model | Accuracy | Precision | Recall | F1-Score |
|---|---:|---:|---:|---:|
| Random Forest | 86.50% | 45.00% | 56.70% | **50.20%** |
| Gradient Boosting | **87.60%** | **48.30%** | 47.80% | 48.00% |

| Logistic Regression | 78.30% | 30.50% | 63.20% | 41.10% |

| Naive Bayes | 69.90% | 24.10% | **70.30%** | 36.00% |
| KNN | 71.9% | 0.245% | 0.645% | 0.355% |

 Best Model Selection

**Random Forest** achieved the highest F1-score of **50.20%**, making it the best overall candidate among the evaluated models.

- **Gradient Boosting** achieved the highest accuracy (87.60%) and precision (48.30%).
- **Naive Bayes** achieved the highest recall (70.30%), identifying the largest proportion of actual subscribers.
- **Random Forest** provided the best balance between precision and recall based on its F1-score.

Therefore, Random Forest is recommended as the leading candidate for further validation before production deployment.

## Challenges Faced and Techniques Used


### 1. Different Feature Scales
- Challenge: Features have different numerical ranges.
- Technique: Used StandardScaler to bring features to a comparable scale.

### 2. Precision and Recall Trade-off
- Challenge: High accuracy does not always mean the model identifies donors correctly.
- Technique: Compared precision, recall and F1-score instead of relying only on accuracy.


### 3. Class Imbalance
- Challenge: Most customers did not subscribe to a term deposit.
- Technique: Used SMOTE to balance the training data and improve the model's ability to identify potential subscribers.

### 4. Unknown Values
- Challenge: Several categorical columns contained unknown values.
- Technique: Analyzed unknown values and handled them appropriately during preprocessing.

### 5. Data Leakage
- Challenge: The duration column is only known after a call and could lead to unrealistic predictions.
- Technique: Excluded duration from the main predictive model.
